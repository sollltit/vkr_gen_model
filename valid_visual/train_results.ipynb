```python
import matplotlib.pyplot as plt
import json
import pandas as pd
```

________________________
## *1 этап обучения:*

```python
with open(r'..\qwen2.5_fine-tune\checkpoint-7341\trainer_state.json', 'r') as f:
    state = json.load(f)

steps = [x["step"] for x in state["log_history"]]
losses = [x["loss"] for x in state["log_history"]]

plt.figure(figsize=(12, 5))
plt.plot(steps, losses, alpha=0.85, label="Training loss", color = '#33cc33')
plt.xlabel("Шаг")
plt.ylabel("loss (потеря)")
plt.title("Кривая обучения (1 этап)")
plt.grid(True, alpha=0.3)
plt.tight_layout()

ma = pd.Series(losses).rolling(window=10).mean()
plt.plot(steps, ma, label="Скользящее среднее", linewidth=2, color = '#ff3399', alpha = 0.65)
plt.legend()
plt.savefig("loss_curve_1.png", dpi=150)
plt.show()
```
